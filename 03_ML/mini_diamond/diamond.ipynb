### M2

```python
# x, y, z 각각 0인 값 개수
print((df[["x", "y", "z"]] == 0).sum())

# 셋 중 하나라도 0인 행
zero_rows = df[(df[["x", "y", "z"]] == 0).any(axis=1)]

# 행 개수
print("0이 포함된 행 개수:", len(zero_rows))

# 해당 행 확인
display(zero_rows)
```
